```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 1000)
```

```python
zero = np.zeros_like(x)

plt.figure(figsize=(6,4))
plt.plot(x, zero, label="Zero MF")
plt.title("Zero Membership Function")
plt.xlabel("x")
plt.ylabel("Membership")
plt.grid(True)
plt.legend()
plt.show()
```

```python
const = np.ones_like(x) * 0.5

plt.figure(figsize=(6,4))
plt.plot(x, const, label="Constant MF")
plt.title("Constant Membership Function")
plt.xlabel("x")
plt.ylabel("Membership")
plt.grid(True)
plt.legend()
plt.show()
```

```python
singleton = np.zeros_like(x)
singleton[np.abs(x-5).argmin()] = 1

plt.figure(figsize=(6,4))
plt.stem(x, singleton)
plt.title("Singleton Membership Function")
plt.xlabel("x")
plt.ylabel("Membership")
plt.grid(True)
plt.show()
```

```python
tri = np.maximum(
    np.minimum((x-2)/(5-2), (8-x)/(8-5)),
    0
)

plt.figure(figsize=(6,4))
plt.plot(x, tri)
plt.title("Triangular Membership Function")
plt.xlabel("x")
plt.ylabel("Membership")
plt.grid(True)
plt.show()
```

```python
ltri = np.where(x<=4,1,np.maximum((8-x)/(8-4),0))

plt.figure(figsize=(6,4))
plt.plot(x,ltri)
plt.title("Left Triangular Membership Function")
plt.xlabel("x")
plt.ylabel("Membership")
plt.grid(True)
plt.show()
```

```python
rtri=np.where(x>=6,1,np.maximum((x-2)/(6-2),0))

plt.figure(figsize=(6,4))
plt.plot(x,rtri)
plt.title("Right Triangular Membership Function")
plt.xlabel("x")
plt.ylabel("Membership")
plt.grid(True)
plt.show()
```

```python
trap = np.maximum(
    np.minimum(
        np.minimum((x-2)/(4-2),1),
        (8-x)/(8-6)
    ),
    0
)

plt.figure(figsize=(6,4))
plt.plot(x,trap)
plt.title("Trapezoidal Membership Function")
plt.xlabel("x")
plt.ylabel("Membership")
plt.grid(True)
plt.show()
```

```python
gauss = np.exp(-((x-5)**2)/(2*(1.2**2)))

plt.figure(figsize=(6,4))
plt.plot(x,gauss)
plt.title("Gaussian Membership Function")
plt.xlabel("x")
plt.ylabel("Membership")
plt.grid(True)
plt.show()
```
